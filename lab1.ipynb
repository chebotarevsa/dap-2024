**8. Что такое Jupyter Notebook, для чего используется?**
****

Jupyter Notebook — это интерактивная среда для программирования и анализа данных, где код, текст, формулы и графики объединяются в одном документе (файле .ipynb).

Используется для:
- написания и выполнения Python-кода по шагам;
- анализа данных и визуализации (pandas, matplotlib и др.);
- создания отчётов и презентаций с кодом;
- обучения и демонстрации алгоритмов машинного обучения.

**9. Почему python популярен для анализа данных?**
****

- Прост в изучении — понятный синтаксис, близкий к человеческому языку;
- Имеет мощные библиотеки — pandas, NumPy, Matplotlib, Seaborn, Scikit-learn;
- Подходит для всех этапов анализа — от очистки данных до машинного обучения;
- Интерактивен — удобно работать в Jupyter Notebook;
- Поддерживает визуализацию и автоматизацию;
- Имеет большое сообщество и множество готовых решений.

**10. Обязательно ли использовать Jupyter Notebook для анализа данных?**
****

Нет, Jupyter Notebook не обязателен для анализа данных — это просто удобный инструмент. Анализ можно выполнять в других средах:
- VS Code, PyCharm — полноценные IDE;
- Google Colab — облачная версия Jupyter;
- Python-скрипты (.py) — подходят для автоматизации;
- RStudio, Spyder, Databricks — альтернативные платформы.

 Однако Jupyter удобен для пошагового анализа, визуализации и презентации результатов.
