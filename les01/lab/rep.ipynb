### 1. Типы ячеек в Jupyter Notebook
Согласно теории, в Jupyter Notebook существуют два основных типа ячеек:
* **Markdown (текстовые) ячейки** — для оформления текста, заголовков, списков и т.д.
* **Python (кодовые) ячейки** — для написания и выполнения программного кода.

### 2. Как с помощью Markdown сформировать иерархический список?
Для создания не нумерованного иерархического (вложенного) списка используются маркеры (`*`, `+` или `-`) с добавлением отступов (пробелов) для вложенных уровней.
*Пример из теории:*
```markdown
* Элемент 1
* Элемент 2
    + Элемент 2 1
    + Элемент 2 2
        + Элемент 2 2 1
        + Элемент 2 2 2
* Элемент 3
```

### 3. Как с помощью Markdown оформить цитату?
Для оформления цитаты необходимо использовать символ `>` в начале строки.
*Пример из теории:*
```markdown
> Если называть Python заменой BASIC, то тогда и трансформер Optimus Prime — это только замена грузовика.
— Cory Dodt
```

### 4. Как в текстовую ячейку вставить отформатированный блок кода?
Необходимо использовать тройные обратные кавычки (```) до и после блока кода, указав язык программирования после первых кавычек (например, `Python`).
*Пример из теории:*
````markdown
```Python
df2 = pd.DataFrame(
    {
        "A": 1.0,
        "B": pd.Timestamp("20130102"),
        ...
    }
)
```
````

### 5. Сколько уровней заголовка существует в Markdown?
Существует **6 уровней** заголовков. Они формируются с помощью символа решетки (`#`), количество которых определяет уровень:
* `#` Заголовок 1 уровня
* `##` Заголовок 2 уровня
* `###` Заголовок 3 уровня
* `####` Заголовок 4 уровня
* `#####` Заголовок 5 уровня
* `######` Заголовок 6 уровня

### 7. Типы данных
**Основные типы данных:**
1. Структурированные данные.
2. Неструктурированные данные.
3. Полуструктурированные данные.

**Подробно про «Структурированные данные»** :
Это тип данных, при котором последние упорядочены в вертикальные столбцы (поля) и горизонтальные строки (записи или наблюдения). Примером таких данных являются данные в реляционных базах и электронных таблицах. Структурированные данные зависят от модели данных, которая представляет собой определенную структуру, содержательного смысла данных и часто от способа обработки данных. Сюда входит идентификация типа данных (целое число, число с плавающей точкой, строка и т.д.) и ограничения, накладываемые на данные (например, количество символов, максимальное и минимальное значения).
