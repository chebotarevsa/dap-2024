### Задача №1. https://leetcode.com/problems/two-sum/

````python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        # Решение
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
        return []
````

### Задача №2. https://leetcode.com/problems/palindrome-number/description/

````python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        # Решение
        if x < 0 or (x % 10 == 0 and x != 0):
            return False

        reversed_half = 0
        while x > reversed_half:
            reversed_half = reversed_half * 10 + x % 10
            x //= 10

        return x == reversed_half or x == reversed_half // 10
````

### Задача №3. https://leetcode.com/problems/roman-to-integer/description/

````python
class Solution(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """
        # Решение
        values = {
            'I': 1, 'V': 5, 'X': 10,
            'L': 50, 'C': 100, 'D': 500, 'M': 1000
        }
        total = 0
        n = len(s)
        for i in range(n):
            if i + 1 < n and values[s[i]] < values[s[i + 1]]:
                total -= values[s[i]]
            else:
                total += values[s[i]]
        return total
````

### Задача №4. https://leetcode.com/problems/longest-common-prefix/description/

````python
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :type strs: List[str]
        :rtype: str
        """
        # Решение
        if not strs:
            return ""

        for i in range(len(strs[0])):
            char = strs[0][i]
            for j in range(1, len(strs)):
                if i >= len(strs[j]) or strs[j][i] != char:
                    return strs[0][:i]
        return strs[0]
````

### Задача №5. https://leetcode.com/problems/merge-two-sorted-lists/description/

````python
class Solution(object):
    def mergeTwoLists(self, list1, list2):
        """
        :type list1: Optional[ListNode]
        :type list2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        # Решение
        dummy = ListNode(0)
        current = dummy

        while list1 and list2:
            if list1.val <= list2.val:
                current.next = list1
                list1 = list1.next
            else:
                current.next = list2
                list2 = list2.next
            current = current.next

        current.next = list1 if list1 else list2
        return dummy.next
````

### Задача №6. https://leetcode.com/problems/longest-substring-without-repeating-characters/description/

````python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        # Решение
        last_seen = {}
        left = 0
        max_len = 0

        for right, char in enumerate(s):
            if char in last_seen and last_seen[char] >= left:
                left = last_seen[char] + 1
            last_seen[char] = right
            max_len = max(max_len, right - left + 1)

        return max_len
````

### 1. Что такое динамическая типизация и чем она отличается от статической?
Динамическая типизация — это свойство языка программирования, при котором тип переменной определяется во время выполнения программы, исходя из значения, которое ей присвоено. Переменная не привязана жёстко к определённому типу: один и тот же идентификатор может в разное время хранить значения разных типов, а проверка корректности типов происходит в момент выполнения операций, а не на этапе компиляции.

Статическая типизация — это свойство, при котором тип переменной фиксируется на этапе компиляции и не может быть изменён во время работы программы. Программист обязан явно указывать тип переменной, а все несоответствия типов выявляются компилятором до запуска программы.

Основные различия:
- Момент определения типа: при динамической типизации — во время выполнения, при статической — во время компиляции.
- Объявление типа: при динамической не требуется, при статической обязательно.
- Смена типа переменной: при динамической разрешена, при статической запрещена.
- Момент обнаружения ошибок типов: при динамической — в рантайме, при статической — на этапе компиляции.
- Скорость работы: динамическая типизация обычно медленнее из-за проверок в рантайме, статическая — быстрее.

### 2. Чем отличается цикл for в Python от цикла for в C++?
В C++ цикл for — это цикл со счётчиком, который состоит из трёх частей: инициализации, условия продолжения и шага. Программист сам управляет индексом и условием выхода, а цикл перебирает числовой диапазон.

В Python цикл for — это цикл обхода коллекции. Он работает по протоколу итератора и перебирает элементы любого итерируемого объекта (списка, строки, словаря, диапазона range и т. д.), не требуя явного управления счётчиком.

Основные различия:
- Назначение: в C++ — счётчик с условием и шагом, в Python — обход элементов коллекции.
- Управление индексом: в C++ явное, в Python неявное (через range, enumerate).
- Что перебирается: в C++ числовой диапазон, в Python любой итерируемый объект.
- Изменение счётчика внутри тела: в C++ влияет на следующую итерацию, в Python не влияет.

### 3. Для чего предназначены операторы continue и break?
Оператор break предназначен для полного прерывания цикла. При его выполнении управление немедленно передаётся первой инструкции, следующей за циклом; оставшиеся итерации не выполняются.

Оператор continue предназначен для прерывания только текущей итерации. При его выполнении управление переходит к следующей итерации цикла: в цикле while — к проверке условия, в цикле for — к следующему элементу коллекции.

Оба оператора применяются для управления ходом выполнения цикла и позволяют избежать лишних проверок и вычислений.

### 4. Как определить функцию в Python? Может ли функция возвращать несколько значений?
Функция в Python определяется с помощью ключевого слова def, после которого указывается имя функции, список параметров в круглых скобках и двоеточие. Тело функции записывается с отступом. Результат возвращается оператором return.

Общая форма определения:
```
python
def имя_функции([параметры]):
    инструкции
    return возвращаемое_значение
```
Функция может возвращать несколько значений. Формально Python возвращает один объект — кортеж (tuple), который синтаксически записывается как несколько значений через запятую. Этот кортеж можно распаковать в отдельные переменные либо использовать как единый объект.

### 5. Что такое лямбда-функции и для чего они предназначены?
Лямбда-функция — это анонимная (безымянная) функция, которая определяется непосредственно в месте использования с помощью ключевого слова lambda. Она содержит только одно выражение, результат которого возвращается автоматически, без использования оператора return.

Лямбда-функции предназначены для коротких операций, которые нужно выполнить один раз или передать как аргумент другой функции. Они особенно удобны при работе с функциями высших порядков (например, sorted, map, filter, max), когда писать полноценную функцию с именем избыточно.

Если логика сложнее одного выражения, лямбда-функция теряет преимущество в компактности, и вместо неё лучше использовать обычную функцию, определённую через def.

### 6. Может ли тело функции содержать определение другой функции?
Да, тело функции может содержать определение другой функции. В Python функции являются объектами первого класса, поэтому определение одной функции внутри другой допустимо и широко используется.

Вложенные функции применяются для:
- инкапсуляции вспомогательной логики, которая не должна быть видна за пределами внешней функции;
- создания замыканий, когда внутренняя функция «запоминает» переменные из области видимости внешней функции;
- построения фабрик функций, возвращающих настроенные функции;
- реализации декораторов — обёрток над другими функциями.

Если внутренняя функция изменяет переменную внешней функции, используется ключевое слово nonlocal, иначе присваивание создаст новую локальную переменную.
