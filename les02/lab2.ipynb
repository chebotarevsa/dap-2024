## Задача №1. Two Sum

```python
class Solution1(object):
  def twoSum(self, nums, target):
      seen_numbers = {}

      for i, num in enumerate(nums):
          complement = target - num

          if complement in seen_numbers:
             return [seen_numbers[complement], i]

          seen_numbers[num] = i

      return []
```

### Проверка 1 задачи

```python
solution = Solution1()
result = solution.twoSum([2, 7, 11, 15], 9)
print("Результат:", result)
```

## Задача №2. Palindrome Number

```python
class Solution2(object):
  def isPalindrome(self, x):

      if x < 0:
           return False
      str_x = str(x)
      return str_x == str_x[::-1]
```

### Проверка 2 задачи

```python
sol2 = Solution2()
print(sol2.isPalindrome(121))
print(sol2.isPalindrome(-121))
print(sol2.isPalindrome(10))
```

## Задача №3. Roman to Integer

```python
class Solution3(object):
  def romanToInt(self, s):
      roman_values = {
          'I': 1, 'V': 5, 'X': 10, 'L': 50,
          'C': 100, 'D': 500, 'M': 1000
      }

      total_sum = 0

      for i, char in enumerate(s):
          current_value = roman_values[char]

          if i + 1 < len(s) and current_value < roman_values[s[i + 1]]:
           total_sum -= current_value
          else:
           total_sum += current_value

      return total_sum
```

### Проверка 3 задачи

```python
sol3 = Solution3()
print(sol3.romanToInt("III"))
print(sol3.romanToInt("LVIII"))
print(sol3.romanToInt("MCMXCIV"))
```

## Задача №4. Longest Common Prefix

```python
class Solution4(object):
    def longestCommonPrefix(self, strs):
        if not strs:
            return ""

        reference_word = strs[0]

        for current_word in strs[1:]:

            while not current_word.startswith(reference_word):
                reference_word = reference_word[:-1]

                if not reference_word:
                    return ""

        return reference_word
```

### Проверка 4 задачи

```python
sol4 = Solution4()

print(sol4.longestCommonPrefix(["flower", "flow", "flight"]))

print(sol4.longestCommonPrefix(["dog", "racecar", "car"]))

print(sol4.longestCommonPrefix(["test", "test", "test"]))

print(sol4.longestCommonPrefix(["ab", "a"]))
```

## Задача №5. Merge Two Sorted Lists

```python
class Solution5(object):
    def mergeTwoLists(self, list1, list2):

        dummy_node = ListNode()
        current_node = dummy_node


        while list1 and list2:

            if list1.val <= list2.val:
                current_node.next = list1
                list1 = list1.next
            else:
                current_node.next = list2
                list2 = list2.next

            current_node = current_node.next


        if list1:
            current_node.next = list1
        else:
            current_node.next = list2


        return dummy_node.next
```

### Проверка 5 задачи

```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def print_list(node):
    result = []
    while node:
        result.append(node.val)
        node = node.next
    print(result)
```

### Проверка 5 задачи

```python
list1 = ListNode(1)
list1.next = ListNode(2)
list1.next.next = ListNode(4)

list2 = ListNode(1)
list2.next = ListNode(3)
list2.next.next = ListNode(4)

sol5 = Solution5()
merged_head = sol5.mergeTwoLists(list1, list2)

print("Результат слияния:")
print_list(merged_head)
```

## Задача №6.

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):

        last_seen_position = {}

        start_of_window = 0
        max_length = 0


        for end_of_window in range(len(s)):
            current_char = s[end_of_window]


            if current_char in last_seen_position and last_seen_position[current_char] >= start_of_window:

                start_of_window = last_seen_position[current_char] + 1


            last_seen_position[current_char] = end_of_window


            current_window_length = end_of_window - start_of_window + 1
            max_length = max(max_length, current_window_length)

        return max_length
```
