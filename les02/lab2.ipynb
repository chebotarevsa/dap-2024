Задача №1. https://leetcode.com/problems/two-sum/

```python
class Solution1(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        seen = {}  # число -> его индекс
        for i, num in enumerate(nums):
            need = target - num
            if need in seen:
                return [seen[need], i]
            seen[num] = i


s = Solution1()
print(s.twoSum([2, 7, 11, 15], 9))  # [0, 1]
print(s.twoSum([3, 2, 4], 6))       # [1, 2]
print(s.twoSum([3, 3], 6))          # [0, 1]
```

Задача №2. https://leetcode.com/problems/palindrome-number/description/

```python
class Solution2(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        s = str(x)
        return s == s[::-1]


s = Solution2()
print(s.isPalindrome(121))   # True
print(s.isPalindrome(-121))  # False
print(s.isPalindrome(10))    # False
```

Задача №3. https://leetcode.com/problems/roman-to-integer/description/

```python
class Solution3(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """
        values = {'I': 1, 'V': 5, 'X': 10, 'L': 50,
                  'C': 100, 'D': 500, 'M': 1000}
        total = 0
        for i in range(len(s)):
            # если следующая цифра больше текущей, текущую вычитаем (IV = 5 - 1)
            if i + 1 < len(s) and values[s[i]] < values[s[i + 1]]:
                total -= values[s[i]]
            else:
                total += values[s[i]]
        return total


s = Solution3()
print(s.romanToInt("III"))      # 3
print(s.romanToInt("LVIII"))    # 58
print(s.romanToInt("MCMXCIV"))  # 1994
```

Задача №4. https://leetcode.com/problems/longest-common-prefix/description/

```python
class Solution4(object):
    def longestCommonPrefix(self, strs):
        """
        :type strs: List[str]
        :rtype: str
        """
        if not strs:
            return ""
        prefix = strs[0]
        for word in strs[1:]:
            # укорачиваем префикс, пока слово не начнётся с него
            while not word.startswith(prefix):
                prefix = prefix[:-1]
        return prefix


s = Solution4()
print(s.longestCommonPrefix(["flower", "flow", "flight"]))  # "fl"
print(s.longestCommonPrefix(["dog", "racecar", "car"]))     # ""
```

Задача №5. https://leetcode.com/problems/merge-two-sorted-lists/description/

```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution5(object):
    def mergeTwoLists(self, list1, list2):
        """
        :type list1: Optional[ListNode]
        :type list2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        dummy = ListNode()  # фиктивный первый узел
        tail = dummy
        while list1 and list2:
            if list1.val <= list2.val:
                tail.next = list1
                list1 = list1.next
            else:
                tail.next = list2
                list2 = list2.next
            tail = tail.next
        tail.next = list1 if list1 else list2  # дописываем остаток
        return dummy.next


# вспомогательные функции для проверки
def to_list(values):
    dummy = ListNode()
    cur = dummy
    for v in values:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def to_python(node):
    result = []
    while node:
        result.append(node.val)
        node = node.next
    return result


s = Solution5()
print(to_python(s.mergeTwoLists(to_list([1, 2, 4]), to_list([1, 3, 4]))))  # [1, 1, 2, 3, 4, 4]
print(to_python(s.mergeTwoLists(to_list([]), to_list([]))))                # []
print(to_python(s.mergeTwoLists(to_list([]), to_list([0]))))               # [0]
```

Задача №6. https://leetcode.com/problems/longest-substring-without-repeating-characters/description/

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        last_seen = {}  # символ -> последний индекс
        start = 0       # начало текущего «окна» без повторов
        best = 0
        for i, ch in enumerate(s):
            if ch in last_seen and last_seen[ch] >= start:
                start = last_seen[ch] + 1  # сдвигаем начало окна за повтор
            last_seen[ch] = i
            best = max(best, i - start + 1)
        return best


s = Solution()
print(s.lengthOfLongestSubstring("abcabcbb"))  # 3
print(s.lengthOfLongestSubstring("bbbbb"))     # 1
print(s.lengthOfLongestSubstring("pwwkew"))    # 3
```
