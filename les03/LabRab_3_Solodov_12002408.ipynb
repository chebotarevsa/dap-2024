**Дан массив 10x2 подеставлющий собой точки в декартовой системе координат, преобразовать в полярную**

```python
import matplotlib.pyplot as plt
nan = [np.nan, np.nan]

points = [
    # --- J ---
    [0.0, 2.0], [1.4, 2.0], [1.4, 0.5], [1.0, 0.0], [0.3, 0.0], [0.0, 0.5],
    nan,
    # --- a : кружок ---
    [1.85, 0.90], [2.10, 1.50], [2.55, 1.70], [3.00, 1.50],
    [3.25, 0.90], [3.00, 0.30], [2.55, 0.10], [2.10, 0.30], [1.85, 0.90],
    nan,
    # --- a : палочка ---
    [3.25, 1.70], [3.25, 0.00],
    nan,
    # --- v ---
    [3.60, 1.70], [4.50, 0.00], [5.40, 1.70],
    nan,
    # --- a : кружок ---
    [5.75, 0.90], [6.00, 1.50], [6.45, 1.70], [6.90, 1.50],
    [7.15, 0.90], [6.90, 0.30], [6.45, 0.10], [6.00, 0.30], [5.75, 0.90],
    nan,
    # --- a : палочка ---
    [7.15, 1.70], [7.15, 0.00],
]

z = np.array(points)
# z = np.random.random((10,2))
print(z)
print("\n")

x = z[0::1,0]
y = z[0::1,1]
print("x=", x, "\n")
print("y=", y, "\n")

r = np.sqrt(x**2 + y**2)
corner = np.arctan2(y,x)

polar = np.column_stack((r, corner))
print("polar = ", polar, "\n")

fig = plt.figure(figsize=(12, 6))

# Декартовы
ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(x, y, 'o-')
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)
ax1.set_title("Декартовы")

# Полярные
ax2 = fig.add_subplot(1, 2, 2, projection='polar')
ax2.plot(corner, r, 'o-')
ax2.set_title("Полярные")

plt.tight_layout()
plt.show()

```
