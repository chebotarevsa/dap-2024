## Ход работы


```python
import numpy as np
```

### Задача №1

Создать массив 10 на 10 со случайными значениями, найти минимум и максимум.


```python
z = np.random.rand(10, 10)
print(z)
print('Минимум:', z.min())
print('Максимум:', z.max())
```

### Задача №2

Создать матрицу 8 на 8 и заполнить её единицами и нулями в шахматном порядке.


```python
z = np.zeros((8, 8), dtype=int)
for i in range(8):
    for j in range(8):
        if (i + j) % 2 == 1:
            z[i, j] = 1
print(z)
```

### Задача №3

Дан массив из 11 элементов, поменять знак у элементов, значения которых между 3 и 8.


```python
z = np.arange(11)
for i in range(len(z)):
    if z[i] > 3 and z[i] < 8:
        z[i] = -z[i]
print(z)
```

### Задача №4

Найти среднее значение каждой строки матрицы и вычесть его из каждого элемента этой строки.


```python
z = np.random.rand(5, 10)
print('До:')
print(z)
for i in range(len(z)):
    average = z[i].mean()
    z[i] = z[i] - average
print('После:')
print(z)
```

### Задача №5

Поменять первую и вторую строки матрицы.


```python
z = np.arange(25).reshape(5, 5)
print('До:')
print(z)
for j in range(5):
    temp = z[0, j]
    z[0, j] = z[1, j]
    z[1, j] = temp
print('После:')
print(z)
```

### Задача №6

Дан массив 10 на 2, представляющий точки в декартовой системе координат. Преобразовать их в полярную систему координат.


```python
z = np.random.random((10, 2))
polar = np.zeros((10, 2))
for i in range(10):
    x = z[i, 0]
    y = z[i, 1]
    polar[i, 0] = np.sqrt(x ** 2 + y ** 2)
    polar[i, 1] = np.arctan2(y, x)
print('Декартовы координаты:')
print(z)
print('Полярные координаты:')
print(polar)
```

### Задача №7

Заменить максимальный элемент массива на ноль.


```python
z = np.random.random(10)
print('До:')
print(z)
max_index = 0
for i in range(len(z)):
    if z[i] > z[max_index]:
        max_index = i
z[max_index] = 0
print('После:')
print(z)
```

### Задача №8

Дан массив, добавить 1 к каждому элементу с индексом, заданным в другом массиве. Добавление производится только 1 раз.


```python
z = np.ones(10)
i = np.random.randint(0, len(z), 20)
print('Индексы:')
print(i)
used = []
for index in i:
    if index not in used:
        z[index] = z[index] + 1
        used.append(index)
print(z)
```

### Задача №9

Дан вектор [1, 2, 3, 4, 5], построить новый вектор с тремя нулями между каждым значением.


```python
z = np.array([1, 2, 3, 4, 5])
result = []
for i in range(len(z)):
    result.append(z[i])
    if i < len(z) - 1:
        for j in range(3):
            result.append(0)
result = np.array(result)
print(result)
```

### Задача №10

Найти ближайшее к заданному значению число в заданном массиве.


```python
z = np.arange(100)
v = np.random.uniform(0, 100)
nearest = z[0]
for number in z:
    if abs(number - v) < abs(nearest - v):
        nearest = number
print('Заданное значение:', v)
print('Ближайшее число:', nearest)
```
