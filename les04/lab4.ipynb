4. Выведите распределение переменной Pclass (социально-экономический статус) и это же распределение, только для мужчин / женщин по отдельности. Сколько было мужчин 2-го класса

```python
print(df["Pclass"].value_counts().sort_index())
print(df[df["Sex"] == "male"]["Pclass"].value_counts().sort_index())
print(df[df["Sex"] == "female"]["Pclass"].value_counts().sort_index())
print("Мужчин 2-го класса:", len(df[(df["Sex"] == "male") & (df["Pclass"] == 2)]))
```

5. Каковы медиана и среднеквадрати́чное отклонение платежей (Fare)? Округлите до 2 десятичных знаков.

```python
print("Медиана:", round(df["Fare"].median(), 2))
print("СКО:", round(df["Fare"].std(), 2))
```

6. Правда ли, что люди моложе 30 лет выживали чаще, чем люди старше 60 лет? Каковы доли выживших в обеих группах?

```python
young = df[df["Age"] < 30]["Survived"].mean()
old = df[df["Age"] > 60]["Survived"].mean()
print("Моложе 30:", round(young * 100, 1), "%")
print("Старше 60:", round(old * 100, 1), "%")
```

Ответ: да, правда. Среди людей моложе 30 лет выжило 40,6%, а среди людей старше 60 лет — 22,7%.

7. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших в обеих группах?

```python
df.groupby("Sex")["Survived"].mean()
```

Ответ: да, правда. Среди женщин выжило 74,2%, а среди мужчин — 18,9%.

8.  Найдите самое популярное имя среди пассажиров Титаника мужского пола?

9. Как отличается средний возраст мужчин / женщин в зависимости от класса обслуживания? Выберите верные утверждения:

    В среднем мужчины 1-го класса старше 40 лет
    В среднем женщины 1-го класса старше 40 лет
    Мужчины всех классов в среднем старше женщин того же класса
    В среднем люди в 1 классе старше, чем во 2-ом, а те старше представителей 3-го класса



```python
df.pivot_table("Age", index="Sex", columns="Pclass", aggfunc="mean")
```

Ответ: верные утверждения:
- В среднем мужчины 1-го класса старше 40 лет (41,3)
- Мужчины всех классов в среднем старше женщин того же класса
- В среднем люди в 1 классе старше, чем во 2-ом, а те старше представителей 3-го класса (38,2 > 29,9 > 25,1)

Неверно: женщины 1-го класса в среднем младше 40 лет (34,6).
