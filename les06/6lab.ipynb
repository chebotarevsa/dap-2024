## Исследование данных

4. Постройте матрицу корреляции.

```python
import matplotlib.pyplot as plt
import seaborn as sns

corr_matrix = df.corr()

print(corr_matrix)

plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Матрица корреляции признаков Diabetes")
plt.show()
```

5. Постройте диаграмму рассеяния целевого значение и признака, коэффициент корреляции которого с  целевым значением, самый высокий.

```python
import matplotlib.pyplot as plt

corr_with_target = df.corr()['Target'].drop('Target')
print("Корреляция признаков с целевой переменной:")
print(corr_with_target)

top_feature = corr_with_target.abs().idxmax()
print(f"\nПризнак с наибольшей корреляцией с Target: {top_feature} (corr={corr_with_target[top_feature]:.2f})")

plt.figure(figsize=(8,6))
plt.scatter(df[top_feature], df['Target'], color='teal', alpha=0.6)
plt.xlabel(top_feature)
plt.ylabel('Target')
plt.title(f'Диаграмма рассеяния Target vs {top_feature}')
plt.grid(True)
plt.show()
```

6. Сформируйте набор признаков (X) из 5 признаков с самым высоким коэффициентом корреляции с целевым значением. Сформируйте набор для целевого значения (y).

```python
corr_with_target = df.corr()['Target'].drop('Target')

top5_features = corr_with_target.abs().sort_values(ascending=False).head(5).index
print("5 признаков с наибольшей корреляцией с Target:")
print(top5_features)

X = df[top5_features]
y = df['Target']

print("\nФорма X:", X.shape)
print("Форма y:", y.shape)
```

## Предсказательная модель

7. Разделите набор данных на два, одни для обучения модели другой для проверки. Тестовый набор должен содержать 25 процентов данных.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Форма X_train:", X_train.shape)
print("Форма X_test:", X_test.shape)
print("Форма y_train:", y_train.shape)
print("Форма y_test:", y_test.shape)
```

8. Выполните обучение модели.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Коэффициенты модели:")
print(model.coef_)
print("\nСвободный член (intercept):", model.intercept_)
print(f"\nMean Squared Error (MSE) на тесте: {mse:.2f}")
print(f"R^2 на тесте: {r2:.2f}")
```

1. Типы машинного обучения:

Обучение с учителем (Supervised Learning): модель обучается на размеченных данных (вход → целевое значение).

Обучение без учителя (Unsupervised Learning): модель ищет структуры в данных без целевых значений (например, кластеризация).

Обучение с частичным контролем (Semi-supervised Learning): комбинация размеченных и неразмеченных данных.

Обучение с подкреплением (Reinforcement Learning): агент обучается на основе наград и штрафов за свои действия.

2. Отличие обучения с учителем и без учителя:

С учителем: есть целевая переменная, задача — предсказать или классифицировать.

Без учителя: целевой переменной нет, задача — найти структуры, группы или закономерности.

3. Отличие пакетного и динамического обучения:

Пакетное (Batch Learning): модель обучается на всем наборе данных разом.

Динамическое (Online/Incremental Learning): модель обновляется по мере поступления новых данных.

4. Отличие обучения на основе образцов и на основе модели:

На основе образцов (Instance-based): модель хранит данные и делает прогноз по близости к новым примерам (например, k-NN).

На основе модели (Model-based): строится обобщённая модель по данным (например, линейная регрессия, нейросеть).

5. Линейная регрессия:

Модель, которая предсказывает числовое значение как линейную комбинацию признаков

6. Градиентный спуск:
Метод оптимизации, используемый для нахождения минимумов функции (например, функции потерь). Модель корректирует параметры по направлению отрицательного градиента.

7. Обработка категориальных признаков:

Label Encoding: присваивание чисел категориям.

One-Hot Encoding: создание бинарных столбцов для каждой категории.

Target Encoding / Frequency Encoding: кодирование на основе статистики целевой переменной.

8. Матрица корреляции:
Таблица, показывающая, как сильно связаны признаки между собой и с целевой переменной. Значения от -1 до 1.

9. Метрика RMSE (Root Mean Squared Error):
Среднеквадратичная ошибка прогнозов, показывает среднее отклонение предсказанных значений от реальных в тех же единицах измерения.

10. Метрика R² (коэффициент детерминации):
Доля дисперсии целевой переменной, объяснённая моделью.

R² = 1 → идеальная модель

R² = 0 → модель не объясняет дисперсию лучше, чем среднее

R² < 0 → модель хуже среднего
