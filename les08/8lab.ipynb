1. Импортируйте модель данных "Ирисы Фишера".

```python
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

print("Признаки:", feature_names)
print("Целевые классы:", target_names)
print("Форма признаков:", X.shape)
print("Форма целевой переменной:", y.shape)
```

2. Выполните нормализацию данных

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_normalized = scaler.fit_transform(X)

print("Первые 5 строк после нормализации:")
print(X_normalized[:5])
```

3. Постройте модель кластеризации используя алгоритм  DBSCAN (https://www.youtube.com/watch?v=svAtnZ5XjSI&t=494s), подберите подходящие параметры eps, min_samples

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import numpy as np

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

dbscan = DBSCAN(eps=0.5, min_samples=5)

labels = dbscan.fit_predict(X_scaled)

print("Метки кластеров:")
print(labels)

plt.figure(figsize=(8,6))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='viridis', s=50)
plt.title("DBSCAN кластеризация Ирисов Фишера")
plt.xlabel("Признак 1 (стандартизированный)")
plt.ylabel("Признак 2 (стандартизированный)")
plt.colorbar(label="Кластеры")
plt.show()
```

4. Постройте диаграмму рассеяния для признаков "sepal length (cm)" "sepal width (cm)", цвет точек на диаграмме должен быть различным для разных кластеров

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import numpy as np

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X[:, :2])

dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X_scaled)

plt.figure(figsize=(8,6))
unique_labels = np.unique(labels)
colors = plt.cm.viridis(np.linspace(0,1,len(unique_labels)))

for k, col in zip(unique_labels, colors):
    class_member_mask = (labels == k)
    xy = X_scaled[class_member_mask]
    if k == -1:
        plt.scatter(xy[:, 0], xy[:, 1], c='red', marker='x', label='Noise')
    else:
        plt.scatter(xy[:, 0], xy[:, 1], c=[col], marker='o', label=f'Cluster {k}')

plt.xlabel('Sepal length (стандартизированный)')
plt.ylabel('Sepal width (стандартизированный)')
plt.title('DBSCAN кластеризация: Sepal Length vs Sepal Width')
plt.legend()
plt.show()
```

5. Сделайте вывод

Вывод: DBSCAN выделил плотные кластеры ирисов и несколько шумовых точек. Кластеры частично совпадают с реальными видами, но перекрытия признаков не позволяют идеально разделить все три класса. Алгоритм хорошо выявляет плотные группы и шум, но для точного разделения лучше использовать классификацию или другие методы кластеризации.

## Вопросы

1. В чем состоит задача кластеризации?
2. Для каких целей используется кластеризацияия?
3. Алгоритм кластеризации Ллойда.
4. Как выбрать количество кластеров для алгоритма K-средних
5. Алгоритм DBSCAN

1. Задача кластеризации:
Разделить набор объектов на группы (кластеры) так, чтобы объекты внутри одного кластера были максимально похожи друг на друга, а объекты разных кластеров — максимально различны.

2. Цели кластеризации:

Выявление скрытых структур и закономерностей в данных.

Сегментация пользователей, клиентов или объектов.

Обнаружение аномалий и шумовых точек.

Сжатие и упрощение данных для последующего анализа.

3. Алгоритм кластеризации Ллойда (K-средних):

Инициализировать K центроидов.

Назначить каждый объект ближайшему центроиду.

Пересчитать центроиды как среднее своих кластеров.

Повторять шаги назначения и пересчета до сходимости (когда центроиды перестают меняться).

4. Выбор числа кластеров для K-средних:

Метод локтя: строится график суммы квадратов расстояний до центроидов для разных K; выбирается точка, где уменьшение ошибки замедляется.

Силуэт (Silhouette score): выбирается K с наибольшим коэффициентом силуэта.

Анализ предметной области: иногда K выбирается исходя из практических требований.

5. Алгоритм DBSCAN:

Находит кластеры по плотности точек.

Основные параметры: eps (радиус окрестности) и min_samples (минимальное число точек для кластера).

Точки, не входящие ни в один плотный кластер, помечаются как шум.

Не требует заранее задавать число кластеров и хорошо работает с шумными данными.
